# Human Face to Emotic Stickers

> A simple tool that takes a human face photo and generates expressive cartoon-style stickers showing emotions like happy, sad, laughing, and crying — all based on the original face. Perfect for chat apps, avatars, and fun personalization.😉"

## The Aim

> For v1.0.0

_The MVP_

 - To Accept a photo containing a clear Human Face
 - Generate cartoon version of that human face, in `happy`, `sad`, `crying` and `laughing` emotions  




# Future Enhancements

  > For v.1.1.0

  _Value Add_

  - Validate the accepted photo for clarity of human face
  - Consider the default emotions as `Happy`, `Sad` and `Cry`. Must be capable to accept and identify other emotions too.
  - Create Sticker pack that are ready to be used in popular IMs like FB Messenger, WhatsApp, Telegram, Line, Viber...
